```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('temp.csv')
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].index
```

```python
for col in missing:
    df[col] = df[col].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.to_csv('cleaned_test.csv')
```

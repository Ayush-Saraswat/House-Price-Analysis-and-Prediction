```python
# Exterior Quality / Condition
quality_map = {
    'Po': 1,
    'Fa': 2,
    'TA': 3,
    'Gd': 4,
    'Ex': 5
}

# Basement Quality / Condition
bsmt_quality_map = {
    'NA': 0,
    'Po': 1,
    'Fa': 2,
    'TA': 3,
    'Gd': 4,
    'Ex': 5
}

# Basement Exposure
bsmt_exposure_map = {
    'NA': 0,
    'No': 1,
    'Mn': 2,
    'Av': 3,
    'Gd': 4
}

# Basement Finish Type
bsmt_finish_map = {
    'NA': 0,
    'Unf': 1,
    'LwQ': 2,
    'Rec': 3,
    'BLQ': 4,
    'ALQ': 5,
    'GLQ': 6
}
```

```python
df['ExterQual'] = df['ExterQual'].map(quality_map)
df['ExterCond'] = df['ExterCond'].map(quality_map)

df['BsmtQual'] = df['BsmtQual'].map(bsmt_quality_map)
df['BsmtCond'] = df['BsmtCond'].map(bsmt_quality_map)

df['BsmtExposure'] = df['BsmtExposure'].map(bsmt_exposure_map)

df['BsmtFinType1'] = df['BsmtFinType1'].map(bsmt_finish_map)
df['BsmtFinType2'] = df['BsmtFinType2'].map(bsmt_finish_map)
```

```python
cols = [
    'ExterQual',
    'ExterCond',
    'BsmtQual',
    'BsmtCond',
    'BsmtExposure',
    'BsmtFinType1',
    'BsmtFinType2'
]

df[cols] = df[cols].fillna(0).astype(int)
```

```python
utilities_map = {
    'ELO': 1,      # electricity only
    'NoSeWa': 2,   # electricity + gas
    'NoSewr': 3,   # electricity + gas + water
    'AllPub': 4    # all public utilities
}
```

```python
df['Utilities'] = (
    df['Utilities']
    .map(utilities_map)
    .astype(int)
)
```

```python
# ==========================================
# TOTAL AREA FEATURES
# ==========================================
df = df.copy()
# Total usable house area
df['TotalArea'] = (
    df['TotalBsmtSF'] +
    df['1stFlrSF'] +
    df['2ndFlrSF']
)

# Total finished living space
df['TotalFinishedSF'] = (
    df['1stFlrSF'] +
    df['2ndFlrSF'] +
    df['BsmtFinSF1'] +
    df['BsmtFinSF2']
)

# Total outdoor area
df['TotalOutdoorArea'] = (
    df['WoodDeckSF'] +
    df['OpenPorchSF'] +
    df['EnclosedPorch'] +
    df['3SsnPorch'] +
    df['ScreenPorch'] +
    df['PoolArea']
)

# ==========================================
# ROOM FEATURES
# ==========================================

# Total rooms
df['TotalRooms'] = (
    df['TotRmsAbvGrd'] +
    df['BedroomAbvGr'] +
    df['KitchenAbvGr']
)

# Average room size
df['RoomDensity'] = (
    df['GrLivArea'] /
    (df['TotRmsAbvGrd'] + 1)
)

# ==========================================
# FLOOR FEATURES
# ==========================================

# Approximate number of floors
df['TotalFloors'] = (
    (df['1stFlrSF'] > 0).astype(int) +
    (df['2ndFlrSF'] > 0).astype(int)
)

# Has second floor
df['Has2ndFloor'] = (
    df['2ndFlrSF'] > 0
).astype(int)

# ==========================================
# BATHROOM FEATURES
# ==========================================

# Total bathrooms
df['TotalBathrooms'] = (
    df['FullBath'] +
    0.5 * df['HalfBath'] +
    df['BsmtFullBath'] +
    0.5 * df['BsmtHalfBath']
)

# Bathrooms per room
df['BathPerRoom'] = (
    df['TotalBathrooms'] /
    (df['TotRmsAbvGrd'] + 1)
)

# Bathroom-quality interaction
df['BathQualityInteraction'] = (
    df['TotalBathrooms'] *
    df['OverallQual']
)

# ==========================================
# KITCHEN FEATURES
# ==========================================

# Total kitchens
df['TotalKitchens'] = df['KitchenAbvGr']

# Kitchen premium
df['KitchenPremium'] = (
    df['KitchenQual'] ** 2
)

# ==========================================
# HOUSE AGE FEATURES
# ==========================================

# Age of house when sold
df['HouseAge'] = (
    df['YrSold'] - df['YearBuilt']
)

# Years since remodel
df['YearsSinceRemodel'] = (
    df['YrSold'] - df['YearRemodAdd']
)

# Whether remodeled
df['IsRemodeled'] = (
    df['YearBuilt'] != df['YearRemodAdd']
).astype(int)

# Recently remodeled
df['RecentRemodel'] = (
    df['YearsSinceRemodel'] <= 10
).astype(int)

# House age buckets
df['HouseAgeBin'] = pd.cut(
    df['HouseAge'],
    bins=[-1, 10, 30, 60, 200],
    labels=[4,3,2,1]
).astype(int)

# ==========================================
# PORCH FEATURES
# ==========================================

# Total porch area
df['TotalPorchArea'] = (
    df['OpenPorchSF'] +
    df['EnclosedPorch'] +
    df['3SsnPorch'] +
    df['ScreenPorch']
)

# Has porch
df['HasPorch'] = (
    df['TotalPorchArea'] > 0
).astype(int)

# Outdoor luxury
df['OutdoorLuxury'] = (
    df['TotalPorchArea'] +
    df['WoodDeckSF'] +
    df['PoolArea']
)

# ==========================================
# POOL FEATURES
# ==========================================

# Has pool
df['HasPool'] = (
    df['PoolArea'] > 0
).astype(int)

# ==========================================
# BASEMENT FEATURES
# ==========================================

# Basement finished ratio
df['BsmtFinishedRatio'] = (
    (df['BsmtFinSF1'] + df['BsmtFinSF2']) /
    (df['TotalBsmtSF'] + 1)
)

# Has finished basement
df['HasFinishedBsmt'] = (
    (df['BsmtFinSF1'] + df['BsmtFinSF2']) > 0
).astype(int)

# Basement score
df['BasementScore'] = (
    df['BsmtQual'] *
    df['BsmtCond'] *
    (df['BsmtExposure'] + 1)
)

# ==========================================
# QUALITY FEATURES
# ==========================================

# Overall quality score
df['OverallScore'] = (
    df['OverallQual'] *
    df['OverallCond']
)

# Exterior score
df['ExteriorScore'] = (
    df['ExterQual'] *
    df['ExterCond']
)

# Luxury score
df['LuxuryScore'] = (
    df['OverallQual'] +
    df['KitchenQual'] +
    df['ExterQual'] +
    df['HeatingQC']
)
# ==========================================
# LIVING SPACE FEATURES
# ==========================================

# Living area ratio
df['LivingAreaRatio'] = (
    df['GrLivArea'] /
    (df['TotalArea'] + 1)
)

# ==========================================
# FIREPLACE FEATURES
# ==========================================
```
